```python
# Data Preprocessing
```

```python
# Separate Input and Output

x = df.drop('Class', axis=1)
y = df['Class']
```

```python
#Train-Test Split
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x, y,
    test_size=0.2,
    random_state=0,
    stratify=y
)
```

```python
# Feature Scaling
```

```python
from sklearn.preprocessing import StandardScaler

sca = StandardScaler()

x_train[['Time', 'Amount']] = sca.fit_transform(x_train[['Time', 'Amount']])

x_test[['Time', 'Amount']] = sca.transform(x_test[['Time', 'Amount']])
```

```python
# SMOTE (Oversampling)
```

```python
from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=0)

x_train_smote, y_train_smote = smote.fit_resample(x_train, y_train)
```

```python
# Random Undersampling
```

```python
from imblearn.under_sampling import RandomUnderSampler

rus = RandomUnderSampler(random_state=0)

x_train_under, y_train_under = rus.fit_resample(x_train, y_train)
```

```python
y_train_under.value_counts()
```

```python
# Logistic Regression (SMOTE)
```

```python
from sklearn.linear_model import LogisticRegression

lr_smote = LogisticRegression(random_state=0)

lr_smote.fit(x_train_smote, y_train_smote)

y_pred_smote = lr_smote.predict(x_test)
```

```python
## Logistic Regression (Random Undersampling)
```

```python
lr_under = LogisticRegression(random_state=0)

lr_under.fit(x_train_under, y_train_under)

y_pred_under = lr_under.predict(x_test)
```

```python
# Logistic Regression (Class Weight)
```

```python
from sklearn.metrics import classification_report, ConfusionMatrixDisplay
```

```python
# Logistic Regression (SMOTE)
```

```python
ConfusionMatrixDisplay.from_predictions(y_test, y_pred_smote)

print(classification_report(y_test, y_pred_smote))
```

```python
# Logistic Regression (Random Undersampling)
```

```python
ConfusionMatrixDisplay.from_predictions(y_test, y_pred_under)

print(classification_report(y_test, y_pred_under))
```

```python
# Logistic Regression (Class Weight)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr_balanced.fit(x_train, y_train)
```

```python
y_pred_balanced = lr_balanced.predict(x_test)
```

```python
ConfusionMatrixDisplay.from_predictions(y_test, y_pred_balanced)

print(classification_report(y_test, y_pred_balanced))
```

```python
from xgboost import XGBClassifier
```

```python
scale_pos_weight = y_train.value_counts()[0] / y_train.value_counts()[1]

print(scale_pos_weight)
```

```python
# Build XGBoost Model
```

```python
xgb = XGBClassifier(
    random_state=0,
    scale_pos_weight=scale_pos_weight,
    eval_metric='logloss'
)

xgb.fit(x_train, y_train)

y_pred_xgb = xgb.predict(x_test)
```

```python
# Business Cost Table
```

```python
import numpy as np

probs = xgb.predict_proba(x_test)[:, 1]
thresholds = np.arange(0.1, 1.0, 0.1)
```

```python
from sklearn.metrics import confusion_matrix
results = []
for t in thresholds:
    preds = (probs >= t).astype(int)
    tn, fp, fn, tp = confusion_matrix(y_test,preds).ravel()
    cost = (fn*10000)+(fp*500)
    results.append([t,tp,fp,fn,cost])
cost_df = pd.DataFrame(results,
    columns=["Threshold","TP","FP","FN","Business Cost"])
cost_df.sort_values("Business Cost").head(10)
```

```python
# Best Threshold
```

```python
best_row = cost_df.loc[cost_df["Business Cost"].idxmin()]
print(best_row)
```

```python
# Observation:
# Best Threshold = 0.10
# Business Cost = ₹137000
# This threshold provides the best trade-off
# between fraud detection and false alarms.
```

```python
# Evaluation 
```

```python
from sklearn.metrics import ConfusionMatrixDisplay, classification_report

ConfusionMatrixDisplay.from_predictions(y_test, y_pred_xgb)

print(classification_report(y_test, y_pred_xgb))
```

```python
# Isolation Forest
```

```python
from sklearn.ensemble import IsolationForest

iso = IsolationForest(contamination=0.0017, random_state=0)

iso.fit(x_train)

y_pred_iso = iso.predict(x_test)

y_pred_iso = np.where(y_pred_iso == -1, 1, 0)
```

```python
from sklearn.metrics import precision_recall_curve, average_precision_score
import matplotlib.pyplot as plt

models = {
    "Logistic": lr_balanced,
    "Random Forest": rf,
    "XGBoost": xgb
}

plt.figure(figsize=(8,6))

for name, model in models.items():

    y_prob = model.predict_proba(x_test)[:,1]

    precision, recall, _ = precision_recall_curve(y_test, y_prob)

    ap = average_precision_score(y_test, y_prob)

    plt.plot(recall, precision, label=f"{name} (AP={ap:.4f})")

plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision-Recall Curve")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Threshold Tuning
```

```python
y_prob = xgb.predict_proba(x_test)[:, 1]
threshold = 0.3

y_pred_threshold = (y_prob >= threshold).astype(int)
ConfusionMatrixDisplay.from_predictions(y_test, y_pred_threshold)

print(classification_report(y_test, y_pred_threshold))
```

```python
# Observation:
# At threshold = 0.30, the XGBoost model achieved 84% Recall and 92% Precision
# for the fraud class. This means the model detected most fraudulent transactions
# while keeping false alarms relatively low. It provides a good balance between
# fraud detection and business cost.
```

```python
# Business Cost Calculation
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred_threshold)

tn, fp, fn, tp = cm.ravel()
```

```python
cost_fp = 500
cost_fn = 10000
total_cost = (fp * cost_fp) + (fn * cost_fn)

print("False Positives :", fp)
print("False Negatives :", fn)
print("Total Business Cost :", total_cost)
```
