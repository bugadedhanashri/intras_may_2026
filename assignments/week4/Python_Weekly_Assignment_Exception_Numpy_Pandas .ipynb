```python
# Q1. Write a function safe_divide(a, b) that divides two numbers and handles ZeroDivisionError and TypeError separately. Use a finally block to print "Operation attempted" regardless of the outcome. Test it with valid numbers, division by zero, and a string input.
```

```python
def safe_divide(a, b):
    try:
        result = a / b
        print("Result:", result)

    except ZeroDivisionError:
        print("Error: Cannot divide by zero.")

    except TypeError:
        print("Error: Invalid data type.")

    finally:
        print("Operation attempted")
safe_divide(10, 2)      # Valid input
safe_divide(10, 0)      # Division by zero
safe_divide(10, "2")    # String input
```

```python
# Q2. Write a function validate_age(age) that raises a custom exception InvalidAgeError if age is negative or greater than 120. Catch this exception and print a user-friendly message. Test it with at least 3 different values.
```

```python
class InvalidAgeError(Exception):
    pass
```

```python
def validate_age(age):
    if age < 0 or age > 120:
        raise InvalidAgeError("Age must be between 0 and 120.")
    print("Valid Age:", age)


# Test Cases
ages = [25, -5, 150]

for age in ages:
    try:
        validate_age(age)
    except InvalidAgeError as e:
        print("Invalid Age Error:", e)
```

```python
#Q3. Write a program that repeatedly asks the user (using a loop with simulated inputs) to enter a number and divide 100 by it. Handle:
##ValueError if input is not a number ZeroDivisionError if input is zero Generic Exception for anything else
```

```python
user_inputs = ['10', 'abc', '0', '5']

for value in user_inputs:
    try:
        num = int(value)
        result = 100 / num
        print(f"100 / {num} = {result}")

    except ValueError:
        print(f"ValueError: '{value}' is not a valid number.")

    except ZeroDivisionError:
        print("ZeroDivisionError: Cannot divide by zero.")

    except Exception as e:
        print("Error:", e)
```

```python
## Q4. Create a 1D NumPy array of integers from 1 to 20 and perform the following operations.
```

```python
import numpy as np

# Create array from 1 to 20
arr = np.arange(1, 21)

# a) Reshape into 4x5 array
arr_2d = arr.reshape(4, 5)
print("4x5 Array:\n", arr_2d)

# b) Sum of rows and columns
print("\nRow Sum:", arr_2d.sum(axis=1))
print("Column Sum:", arr_2d.sum(axis=0))

# c) Elements greater than 10
print("\nElements > 10:", arr[arr > 10])

# d) Replace even numbers with -1
modified = arr.copy()
modified[modified % 2 == 0] = -1
print("\nModified Array:", modified)
```

```python
# Q5. Create two NumPy arrays A and B and perform operations.
```

```python
import numpy as np

A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])

# a) Element-wise operations
print("Addition:\n", A + B)
print("\nSubtraction:\n", A - B)
print("\nMultiplication:\n", A * B)

# b) Matrix Multiplication
print("\nMatrix Multiplication:\n", np.dot(A, B))

# c) Transpose of A
print("\nTranspose of A:\n", A.T)

# d) Determinant and Inverse
print("\nDeterminant of A:", np.linalg.det(A))
print("\nInverse of A:\n", np.linalg.inv(A))
```

```python
# Q6. Generate a NumPy array of 15 random integers between 1 and 100 (seed=42).
```

```python
import numpy as np

np.random.seed(42)

arr = np.random.randint(1, 101, 15)

print("Array:", arr)

# a) Statistics
print("Mean:", np.mean(arr))
print("Median:", np.median(arr))
print("Standard Deviation:", np.std(arr))
print("Minimum:", np.min(arr))
print("Maximum:", np.max(arr))

# b) Descending Order
print("\nDescending Order:")
print(np.sort(arr)[::-1])

# c) Indices of 3 largest values
indices = np.argsort(arr)[-3:]
print("\nIndices of 3 Largest Values:", indices)

# d) Normalize values between 0 and 1
normalized = (arr - np.min(arr)) / (np.max(arr) - np.min(arr))
print("\nNormalized Array:")
print(normalized)
```

```python
# Q7. Create a Pandas DataFrame from the dictionary below and perform the following.
```

```python
import pandas as pd
data = {
    "Name": ["Aarav", "Priya", "Karan", "Meera", "Sahil"],
    "Age": [23, 21, 25, 22, 24],"Department": ["AI", "Data Science", "AI", "Web Dev", "Data Science"],
    "Score": [88, 92, 79, 95, 67]
}
df = pd.DataFrame(data)

# a) First 3 rows and data types
print(df.head(3))
print(df.dtypes)

# b) Add Grade column
df["Grade"] = df["Score"].apply(
    lambda x: "A" if x >= 90 else ("B" if x >= 75 else "C")
)

# c) Sort by Score descending
print(df.sort_values(by="Score", ascending=False))

# d) Filter Department = AI
print(df[df["Department"] == "AI"])
```

```python
# Q9. Handle Missing Values
#Question Data data = {"Product": ["Pen", "Notebook", "Eraser", "Sharpener", "Ruler"],
#"Price": [10, np.nan, 5, 8, np.nan],"Stock": [100, 50, np.nan, 30, 20]}
```

```python
import pandas as pd
import numpy as np

data = {
    "Product": ["Pen", "Notebook", "Eraser", "Sharpener", "Ruler"],
    "Price": [10, np.nan, 5, 8, np.nan],
    "Stock": [100, 50, np.nan, 30, 20]
}

df = pd.DataFrame(data)

# a) Missing values count
print(df.isnull().sum())

# b) Fill Price with mean
df["Price"].fillna(df["Price"].mean(), inplace=True)

# c) Drop rows where Stock is missing
df.dropna(subset=["Stock"], inplace=True)

# d) Save to CSV
df.to_csv("cleaned_products.csv", index=False)

print(df)
```
